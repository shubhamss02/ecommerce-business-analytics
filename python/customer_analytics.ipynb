```python
primary_keys = {
    'users': (users, 'user_id'),
    'orders': (orders, 'order_id'),
    'order_items': (order_items, 'order_item_id'),
    'products': (products, 'product_id'),
    'reviews': (reviews, 'review_id'),
    'events': (events, 'event_id')
}

for table_name, (df, key) in primary_keys.items():
    print(
        table_name,
        '| unique:', df[key].is_unique,
        '| missing:', df[key].isna().sum()
    )
```

```python
foreign_keys = [
    ('orders.user_id', orders['user_id'], users['user_id']),
    ('order_items.order_id', order_items['order_id'], orders['order_id']),
    ('order_items.product_id', order_items['product_id'], products['product_id']),
    ('order_items.user_id', order_items['user_id'], users['user_id']),
    ('reviews.order_id', reviews['order_id'], orders['order_id']),
    ('reviews.product_id', reviews['product_id'], products['product_id']),
    ('reviews.user_id', reviews['user_id'], users['user_id']),
    ('events.user_id', events['user_id'], users['user_id']),
    ('events.product_id', events['product_id'], products['product_id'])
]

for relationship, child_key, parent_key in foreign_keys:
    invalid_count = (~child_key.isin(parent_key)).sum()
    print(relationship, '| invalid values:', invalid_count)
```

```python
from sqlalchemy import create_engine
from urllib.parse import quote_plus

password = quote_plus("password")

engine = create_engine(
    f"postgresql+psycopg2://postgres:{password}@localhost:5432/ecommerce"
)
```

```python
tables = {
    "users": users,
    "orders": orders,
    "order_items": order_items,
    "products": products,
    "reviews": reviews,
    "events": events
}

for table_name, df in tables.items():
    df.to_sql(
        table_name,
        engine,
        if_exists="replace",
        index=False
    )
    print(f"{table_name} uploaded successfully!")
```
